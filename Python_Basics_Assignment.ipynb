Question 1 : What is Python and why is it widely used in Data Analytics? Mention any three reasons.

Answer 1 :

Python is a high-level, interpreted, general-purpose programming language known for its simple, readable syntax that resembles plain English. It supports multiple programming styles (procedural, object-oriented, and functional) and has a massive ecosystem of libraries, which is why it has become one of the most widely used languages in Data Analytics.

Three reasons Python is widely used in Data Analytics:
     
*   Rich ecosystem of libraries: Libraries like Pandas, NumPy, Matplotlib, and Seaborn provide ready-made tools for data cleaning, numerical computation, and visualization, so analysts don't need to write everything from scratch.
*   Easy to learn and read: Python's simple, English-like syntax means analysts (even those without a heavy programming background) can pick it up quickly and write clear, maintainable code.
*   Strong community and integration: Python integrates well with databases, big data tools (like Spark), and machine learning frameworks (like Scikit-learn and TensorFlow), and has a large community providing support, tutorials, and constant library updates.

Question 2 : Explain the difference between List and Tuple in Python.

Answer :
1. **Give the syntax difference first** — List uses [ ], Tuple uses ( ).
2. **Explain mutability** — this is the core difference: List is mutable (can change), Tuple is immutable (cannot change).
3. **Mention performance** — Tuples are slightly faster since they're fixed-size.
4. **Give a use-case example** — List for data that changes, Tuple for fixed data like coordinates.
5. **Optional: mention methods** — Lists have more built-in methods (append, remove, sort) since they support modification.

Question 3 : What is a function in Python? Why are functions useful?

Answer:

A function in Python is a named, reusable block of code that performs a specific task. It is defined once using the def keyword and can be called (executed) as many times as needed, optionally accepting inputs (parameters) and returning an output (a return value).

Functions are useful because:


*   Reusability: Once written, a function can be called multiple times from different parts of a program instead of rewriting the same code repeatedly.
*   Organization and readability: Breaking a program into functions makes the code easier to read, understand, and maintain, since each function handles one clear task.
*   Easier debugging and testing: Since each function is self-contained, it's easier to test and fix issues in a specific function without affecting the rest of the program.

Question 5 : Write a Python program to check whether a number is even or odd using conditional
statements.

```python
number = int(input("Enter a number: "))
if number % 2 == 0:
    print(f"{number} is an Even number.")
else:
    print(f"{number} is an odd number.")
```

Question 8 : Write a Python program to remove duplicate values from a list using a set.

```python
raw_list = [10, 25, 10, 40, 50, 25, 60]

unique_list = list(set(raw_list))

print("Original list with duplicates:", raw_list)
print("Cleaned list without duplicates:", unique_list)
```

Question 9 : Write a function that returns the square of a number.

```python
def find_square(number):
    return number ** 2

number = 8
squared_result = find_square(number)

print(f"The square of {number} is: {squared_result}")
```

Question 10 : Write a Python program to count how many times a number appears in a list.

Example List

[2,3,4,2,5,2]

```python
def count_occurrences(lst, num):
    return lst.count(num)

example_list = [2, 3, 4, 2, 5, 2]
number_to_count = 2
result = count_occurrences(example_list, number_to_count)
print(f"{number_to_count} appears {result} times in the list.")
```
